* **題目說明**：本程式負責處理含有文字格式的原始餐費與人數資料，進行字串清理、型別轉換，並依照「計算服務費 -> 扣除折價券 -> 平均分攤 -> 判斷是否超支」的規則計算出聚餐各項費用與預算狀態。

```python
# ==========================================
# 測試一：使用原始資料（4 人、折價券 100 元）
# ==========================================
print("=== 測試一執行結果 ===")
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
discount = 100
budget_per_person = 300

# 1. 整理聚餐名稱：去除頭尾空白並轉為小寫
cleaned_event_name = event_name.strip().lower()

# 2. 整理餐費與人數並轉換型別
# 移除 "NT$", 逗號與空白，並將字串轉成浮點數
cleaned_meal_amount = float(meal_amount.replace("NT$", "").replace(",", "").strip())
# 將人數文字轉成整數
num_people = int(people)

# 3. 分步計算費用
service_fee = cleaned_meal_amount * service_rate
total_amount = cleaned_meal_amount + service_fee - discount
cost_per_person = total_amount / num_people

# 4. 判斷是否超過預算（回傳布林值）
is_over_budget = cost_per_person > budget_per_person

# 5. 輸出費用摘要 (使用 f-string 將金額格式化到小數點後兩位)
print(f"聚餐名稱：{cleaned_event_name}")
print(f"原始餐費：{cleaned_meal_amount:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{num_people} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print(f"是否超過預算：{is_over_budget}\n")
```
